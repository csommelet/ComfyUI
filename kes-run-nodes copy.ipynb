```python
import torch
import math
import numpy as np

def print_tensor_closeness(a: torch.Tensor, b: torch.Tensor, *, atol=1e-5, rtol=1e-3, name_a="a", name_b="b"):
    assert a.shape == b.shape, f"shape mismatch: {a.shape} vs {b.shape}"
    with torch.no_grad():
        a = a.detach()
        b = b.detach()
        mask = torch.isfinite(a) & torch.isfinite(b)
        n_total = a.numel()
        n_used = int(mask.sum().item())
        if n_used == 0:
            print("No finite elements to compare.")
            return

        ad = (a - b)[mask].float()
        aa = a[mask].float().abs()
        ab = b[mask].float().abs()
        denom = torch.maximum(aa, ab)

        max_abs = ad.abs().max().item()
        mean_abs = ad.abs().mean().item()
        rms = torch.sqrt((ad * ad).mean()).item()
        mean_rel = (ad.abs() / torch.clamp(denom, min=1e-12)).mean().item()

        # cosine similarity on flattened vectors
        cos = torch.nn.functional.cosine_similarity(
            a[mask].reshape(-1), b[mask].reshape(-1), dim=0
        ).item()

        # per-element allclose using PyTorch criterion
        tol = atol + rtol * denom
        n_fail = int((ad.abs() > tol).sum().item())
        frac_fail = n_fail / n_used

        overall_allclose = torch.allclose(a, b, atol=atol, rtol=rtol, equal_nan=False)

        print(f"{name_a} vs {name_b}")
        print(f"  elements: {n_used}/{n_total} finite")
        print(f"  allclose(atol={atol}, rtol={rtol}): {overall_allclose}")
        print(f"  failures: {n_fail}  ({frac_fail:.4%})")
        print(f"  max|a-b|: {max_abs:.6g}")
        print(f"  mean|a-b|: {mean_abs:.6g}")
        print(f"  RMS error: {rms:.6g}")
        print(f"  mean relative error: {mean_rel:.6g}")
        print(f"  cosine similarity: {cos:.6g}")
```

```python
from safetensors import safe_open
import re
from collections import Counter

# heuristics: adapt patterns to your model family
TEXT_PATTERNS = [
    r"\btext\b", r"\bclip\b", r"\bt5\b", r"\bencoder\b",
    r"text_model", r"transformer", r"token_",
    r"text_enc", r"text_encoder", r"prompt_encoder",
]
UNET_PATTERNS = [r"\bunet\b", r"input_blocks", r"middle_block", r"output_blocks"]

def check_lora_targets(path):
    mods = []
    with safe_open(path, framework="pt", device="cpu") as f:
        for k in f.keys():
            mods.append(k)
    def match_any(key, pats): return any(re.search(p, key) for p in pats)

    hits_text = [k for k in mods if match_any(k.lower(), TEXT_PATTERNS)]
    hits_unet = [k for k in mods if match_any(k.lower(), UNET_PATTERNS)]
    top_prefix = Counter(k.split(".")[0] for k in mods).most_common(10)


    if hits_text:
        print("=====", path)
        print(f"tensors: {len(mods)}")
        print("top-level prefixes:", top_prefix)
        print(f"text-like tensors: {len(hits_text)}")
        print(f"unet-like tensors: {len(hits_unet)}")
        print("→ This LoRA has tensors mapped to the text/CLIP encoder.")
    # else:
    #     print("→ No obvious text/CLIP tensors detected (likely UNet-only).")
```

```python
import os
for dirpath, _, filenames in os.walk("/workspace/ComfyUI/models/loras"):
    for fn in filenames:
        if fn.lower().endswith(".safetensors"):
            check_lora_targets(os.path.join(dirpath, fn))
```
